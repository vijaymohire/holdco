# Phase 4 — Open-Loop Simulation / QAI Evaluation

This section is appended to the Phase 3 notebook to preserve notebook lineage.

**Phase 3 remains the workflow-definition baseline.**

Phase 4 extends that workflow into a controlled open-loop experiment:

**Input / Sensing → Processing → Digital Farm Twin Context → Classical / AI / QAI → Optimisation → Recommendation → KPI / Value Measurement → Evidence**

**Pilot boundary:** no automated physical actuation.


## 12. Phase 4 Experiment Configuration

Define one bounded agriculture optimisation problem and its controlled experiment configuration.

The initial implementation uses synthetic data so that the experiment is deterministic and reproducible. Real or historical data can be introduced later without changing the logical workflow.


```python
# Phase 4 — experiment configuration
from dataclasses import dataclass, asdict
from typing import Dict, Any, List
import json
import math
import statistics

@dataclass
class Phase4Config:
    experiment_id: str = "FAEP-P4-OPENLOOP-001"
    crop: str = "vegetable"
    field_area_ha: float = 10.0
    scenario: str = "cold_weather_water_constraint"
    data_realism: str = "synthetic"
    device_realism: str = "virtual"
    actuation_mode: str = "recommendation_only"
    random_seed: int = 42

config = Phase4Config()

print("Phase 4 configuration")
for k, v in asdict(config).items():
    print(f"  {k}: {v}")
```

## 13. Input State / Scenario

The experiment represents a bounded field condition using synthetic sensing and environmental inputs.

The scenario is deliberately small so that the complete sensing-to-recommendation path can be inspected before adding physical assets.


```python
# Phase 4 — synthetic sensing/environment input
sensor_input = {
    "soil_moisture_pct": 24.0,
    "air_temperature_c": -3.0,
    "relative_humidity_pct": 82.0,
    "water_available_l": 18000.0,
    "crop_stress_index": 0.35,
    "equipment_availability_pct": 90.0,
    "inventory_units": 1200.0,
    "demand_units": 1100.0,
    "transport_capacity_units": 1150.0,
}

environment = {
    "frost_risk": 0.72,
    "precipitation_mm": 2.0,
    "forecast_cold_hours": 8.0,
}

print("Synthetic input state:")
for k, v in sensor_input.items():
    print(f"  {k}: {v}")

print("\nEnvironmental scenario:")
for k, v in environment.items():
    print(f"  {k}: {v}")
```

## 14. Sensing / Input Validation

Validate the experiment inputs before constructing the Digital Farm Twin context.


```python
# Phase 4 — input validation
required_inputs = [
    "soil_moisture_pct",
    "air_temperature_c",
    "relative_humidity_pct",
    "water_available_l",
    "crop_stress_index",
    "equipment_availability_pct",
    "inventory_units",
    "demand_units",
    "transport_capacity_units",
]

missing = [k for k in required_inputs if k not in sensor_input]
range_checks = {
    "soil_moisture_pct": 0 <= sensor_input["soil_moisture_pct"] <= 100,
    "relative_humidity_pct": 0 <= sensor_input["relative_humidity_pct"] <= 100,
    "crop_stress_index": 0 <= sensor_input["crop_stress_index"] <= 1,
    "equipment_availability_pct": 0 <= sensor_input["equipment_availability_pct"] <= 100,
}

input_validation = {
    "missing_inputs": missing,
    "range_checks": range_checks,
    "valid": (len(missing) == 0 and all(range_checks.values()))
}

print("Input validation:", input_validation)
```

## 15. Digital Farm Twin Context

Construct the minimum virtual system state required for the experiment.

This is an experimental Twin context, not a claim of a production Digital Twin.


```python
# Phase 4 — minimum Digital Farm Twin context
twin_context = {
    "field": {
        "id": "FIELD-001",
        "area_ha": config.field_area_ha,
        "crop": config.crop,
    },
    "environment": environment,
    "sensing": sensor_input,
    "operational": {
        "equipment_availability_pct": sensor_input["equipment_availability_pct"],
        "inventory_units": sensor_input["inventory_units"],
        "demand_units": sensor_input["demand_units"],
        "transport_capacity_units": sensor_input["transport_capacity_units"],
    },
    "data_realism": config.data_realism,
    "device_realism": config.device_realism,
}

print("Digital Farm Twin context constructed.")
print(json.dumps(twin_context, indent=2))
```

## 16. Classical Baseline

Establish a transparent Classical baseline before evaluating an alternative computational path.

For this first open-loop slice, the baseline selects a simple rule-based recommendation from the observed state. It is intentionally interpretable rather than presented as an optimal production algorithm.


```python
# Phase 4 — transparent Classical baseline
def classical_baseline(state: Dict[str, Any]) -> Dict[str, Any]:
    actions = []
    if state["soil_moisture_pct"] < 30 and state["water_available_l"] > 5000:
        actions.append("prioritise_water_allocation")
    if state["frost_risk"] >= 0.60:
        actions.append("activate_frost_protection_plan")
    if state["inventory_units"] < state["demand_units"]:
        actions.append("prioritise_supply_recovery")

    return {
        "method": "classical_rule_baseline",
        "actions": actions,
        "feasible": True,
    }

baseline_result = classical_baseline(sensor_input)
print(json.dumps(baseline_result, indent=2))
```

## 17. AI / QAI-Compatible Alternative

The Phase 4 architecture permits AI/QAI alternatives, but this initial notebook uses a small deterministic scoring model as a **QAI-compatible experimental placeholder** rather than claiming quantum execution or quantum advantage.

The important Phase 4 objective is to establish the common comparison interface.


```python
# Phase 4 — deterministic alternative scoring model
# This is NOT a quantum-advantage claim and does not require a QPU.

candidate_actions = [
    {"name": "maintain_current_plan", "water_use_l": 0, "risk_reduction": 0.05, "cost": 0},
    {"name": "prioritise_water_allocation", "water_use_l": 3000, "risk_reduction": 0.35, "cost": 120},
    {"name": "activate_frost_protection_plan", "water_use_l": 1000, "risk_reduction": 0.45, "cost": 180},
    {"name": "combined_water_and_frost_plan", "water_use_l": 4000, "risk_reduction": 0.65, "cost": 260},
]

def score_candidate(candidate, state, env):
    risk_weight = 1000.0
    resource_weight = 0.01
    return (
        risk_weight * candidate["risk_reduction"] * (0.5 + env["frost_risk"])
        - resource_weight * candidate["water_use_l"]
        - candidate["cost"]
        - 200.0 * state["crop_stress_index"]
    )

scored_candidates = [
    {**c, "score": score_candidate(c, sensor_input, environment)}
    for c in candidate_actions
]

alternative_result = max(scored_candidates, key=lambda x: x["score"])

print("Alternative computational path:")
print(json.dumps(alternative_result, indent=2))
```

## 18. Common Evaluation

Classical and alternative paths are evaluated using a common interface.

This is the point at which later AI, hybrid QAI, or optional remote quantum implementations can replace the placeholder without changing the experiment contract.


```python
# Phase 4 — common evaluation
classical_actions = set(baseline_result["actions"])

def evaluate_result(result_name, recommendation, estimated_cost):
    return {
        "method": result_name,
        "recommendation": recommendation,
        "estimated_cost": estimated_cost,
        "execution_status": "completed",
    }

classical_eval = evaluate_result(
    "Classical baseline",
    baseline_result["actions"],
    120 if baseline_result["actions"] else 0
)

alternative_eval = evaluate_result(
    "Alternative scoring path",
    [alternative_result["name"]],
    alternative_result["cost"]
)

comparison = {
    "classical": classical_eval,
    "alternative": alternative_eval,
}

print(json.dumps(comparison, indent=2))
```

```python
# Phase 4 — recommendation boundary
recommendation = {
    "experiment_id": config.experiment_id,
    "recommended_action": alternative_result["name"],
    "reason": "highest experimental score under the configured scenario",
    "estimated_cost": alternative_result["cost"],
    "estimated_water_use_l": alternative_result["water_use_l"],
    "risk_reduction_score": alternative_result["risk_reduction"],
    "control_mode": "human_review_required",
    "physical_actuation": False,
}

print("OPEN-LOOP RECOMMENDATION")
print(json.dumps(recommendation, indent=2))
```

# Phase 5 — Closed-Loop CPS Simulation

This section is appended to the Phase 4 Open-Loop notebook.

**Phase 4 remains the open-loop experimental baseline.**

Phase 5 extends the same experiment into a controlled closed-loop simulation:

**Observe → Process → Decide → Authorise → Actuate → Changed State → Feedback → Next Cycle**

Pilot boundary: actuation is simulated/emulated. No automated physical actuator is invoked.


## 26. Phase 5 Experiment Configuration

Define a controlled closed-loop experiment using the Phase 4 state, scenario and decision concepts.

The initial experiment remains synthetic and deterministic so that the complete cycle can be reproduced and inspected.


```python
# Phase 5 — closed-loop configuration
from dataclasses import dataclass, asdict
from typing import Dict, Any, List
import copy
import json

@dataclass
class Phase5Config:
    experiment_id: str = "FAEP-P5-CLOSEDLOOP-001"
    max_cycles: int = 5
    data_realism: str = "synthetic"
    device_realism: str = "emulated"
    actuation_mode: str = "simulated_emulated"
    human_approval_mode: str = "simulated_approval"
    random_seed: int = 42

phase5_config = Phase5Config()

print("Phase 5 configuration")
for k, v in asdict(phase5_config).items():
    print(f"  {k}: {v}")
```

## 27. Initial Closed-Loop State

Start from the controlled Phase 4 state representation.

The state is copied rather than modified in place so the Phase 4 baseline remains conceptually preserved.


```python
# Phase 5 — initialise closed-loop state from the Phase 4 experiment
closed_loop_state = copy.deepcopy(sensor_input)

print("Initial closed-loop state:")
print(json.dumps(closed_loop_state, indent=2))
```

## 28. Decision Function

Reuse the Phase 4 decision concepts, but make the output explicit as a CPS decision object.

The decision is separate from authorization and actuation.


```python
# Phase 5 — decision layer
def make_decision(state: Dict[str, Any], env: Dict[str, Any]) -> Dict[str, Any]:
    if state["soil_moisture_pct"] < 30 and env["frost_risk"] >= 0.60:
        action = "combined_water_and_frost_plan"
    elif env["frost_risk"] >= 0.60:
        action = "activate_frost_protection_plan"
    elif state["soil_moisture_pct"] < 30:
        action = "prioritise_water_allocation"
    else:
        action = "maintain_current_plan"

    return {
        "decision": action,
        "reason": "controlled rule-based decision for Phase 5 baseline",
        "state_snapshot": copy.deepcopy(state),
    }

decision = make_decision(closed_loop_state, environment)
print(json.dumps(decision, indent=2))
```

## 29. Authorization Boundary

Authorization is explicitly separated from the decision.

For this controlled simulation, approval is represented as a configurable simulated human-approval step.


```python
# Phase 5 — authorization layer
def authorize_decision(decision: Dict[str, Any], approval_mode: str) -> Dict[str, Any]:
    approved = approval_mode == "simulated_approval"
    return {
        "decision": decision["decision"],
        "authorized": approved,
        "authorization_mode": approval_mode,
        "reason": "approved within controlled simulation" if approved else "not authorized",
    }

authorization = authorize_decision(
    decision,
    phase5_config.human_approval_mode
)

print(json.dumps(authorization, indent=2))
```

## 30. Simulated / Emulated Actuation

The actuation layer changes only the simulated state.

No physical device, PLC, pump, valve, drone or other actuator is called.


```python
# Phase 5 — simulated/emulated actuation
def apply_simulated_actuation(
    state: Dict[str, Any],
    action: str,
    authorized: bool
) -> Dict[str, Any]:

    next_state = copy.deepcopy(state)

    if not authorized:
        return {
            "state": next_state,
            "actuated": False,
            "action": action,
            "message": "Action not authorized; state unchanged."
        }

    if action == "combined_water_and_frost_plan":
        next_state["soil_moisture_pct"] = min(
            100.0, next_state["soil_moisture_pct"] + 8.0
        )
        next_state["water_available_l"] = max(
            0.0, next_state["water_available_l"] - 4000.0
        )
        next_state["crop_stress_index"] = max(
            0.0, next_state["crop_stress_index"] - 0.12
        )

    elif action == "activate_frost_protection_plan":
        next_state["crop_stress_index"] = max(
            0.0, next_state["crop_stress_index"] - 0.10
        )

    elif action == "prioritise_water_allocation":
        next_state["soil_moisture_pct"] = min(
            100.0, next_state["soil_moisture_pct"] + 6.0
        )
        next_state["water_available_l"] = max(
            0.0, next_state["water_available_l"] - 3000.0
        )

    return {
        "state": next_state,
        "actuated": True,
        "action": action,
        "message": "Simulated/emulated actuation applied."
    }

actuation = apply_simulated_actuation(
    closed_loop_state,
    authorization["decision"],
    authorization["authorized"]
)

print(json.dumps(actuation, indent=2))
```

## 31. Changed State

The result of the simulated actuation becomes the new represented system state.

This is the critical transition from decision to changed CPS state.


```python
# Phase 5 — changed state
changed_state = actuation["state"]

print("Changed state:")
print(json.dumps(changed_state, indent=2))
```

## 32. Sensing / Feedback

Generate the next observation from the changed state.

The feedback function is deliberately explicit so that later physical or real-data interfaces can replace the simulation without changing the closed-loop contract.


```python
# Phase 5 — sensing feedback
def sense_feedback(state: Dict[str, Any]) -> Dict[str, Any]:
    return {
        "soil_moisture_pct": state["soil_moisture_pct"],
        "water_available_l": state["water_available_l"],
        "crop_stress_index": state["crop_stress_index"],
        "equipment_availability_pct": state["equipment_availability_pct"],
        "inventory_units": state["inventory_units"],
        "demand_units": state["demand_units"],
        "transport_capacity_units": state["transport_capacity_units"],
    }

feedback = sense_feedback(changed_state)

print("Feedback observation:")
print(json.dumps(feedback, indent=2))
```

## 33. Closed-Loop Cycle

Execute repeated controlled cycles.

Each cycle records:

**Observation → Decision → Authorization → Actuation → Changed State → Feedback**

No physical actuation is performed.


```python
# Phase 5 — repeated closed-loop simulation
cycle_records = []
current_state = copy.deepcopy(closed_loop_state)

for cycle_number in range(1, phase5_config.max_cycles + 1):

    observation = sense_feedback(current_state)
    cycle_decision = make_decision(observation, environment)

    cycle_authorization = authorize_decision(
        cycle_decision,
        phase5_config.human_approval_mode
    )

    cycle_actuation = apply_simulated_actuation(
        current_state,
        cycle_authorization["decision"],
        cycle_authorization["authorized"]
    )

    next_state = cycle_actuation["state"]
    next_feedback = sense_feedback(next_state)

    cycle_record = {
        "cycle": cycle_number,
        "observation": observation,
        "decision": cycle_decision,
        "authorization": cycle_authorization,
        "actuation": {
            "action": cycle_actuation["action"],
            "actuated": cycle_actuation["actuated"],
        },
        "changed_state": next_state,
        "feedback": next_feedback,
    }

    cycle_records.append(cycle_record)
    current_state = next_state

print(f"Completed closed-loop cycles: {len(cycle_records)}")
for record in cycle_records:
    print(
        f"Cycle {record['cycle']}: "
        f"{record['decision']['decision']} -> "
        f"actuated={record['actuation']['actuated']}"
    )
```

## 34. State-Transition Verification

Verify that each authorized simulated action produces an observable state result.

An unchanged state can also be valid when the selected action is intentionally non-state-changing, but the result must be explicit.


```python
# Phase 5 — state transition verification
transition_checks = []

for record in cycle_records:
    before = record["observation"]
    after = record["changed_state"]

    changed = any(
        before.get(key) != after.get(key)
        for key in [
            "soil_moisture_pct",
            "water_available_l",
            "crop_stress_index",
            "inventory_units",
        ]
    )

    transition_checks.append({
        "cycle": record["cycle"],
        "authorized": record["authorization"]["authorized"],
        "state_change_observed": changed,
        "explicit_result": True,
    })

print(json.dumps(transition_checks, indent=2))
```

## 35. Feedback Integrity

Confirm that the feedback used by the next cycle corresponds to the state produced by the previous cycle.


```python
# Phase 5 — feedback integrity
feedback_integrity = []

for i in range(len(cycle_records) - 1):
    current_after = cycle_records[i]["changed_state"]
    next_observation = cycle_records[i + 1]["observation"]

    consistent = all(
        current_after.get(key) == next_observation.get(key)
        for key in [
            "soil_moisture_pct",
            "water_available_l",
            "crop_stress_index",
            "equipment_availability_pct",
            "inventory_units",
            "demand_units",
            "transport_capacity_units",
        ]
    )

    feedback_integrity.append({
        "from_cycle": cycle_records[i]["cycle"],
        "to_cycle": cycle_records[i + 1]["cycle"],
        "feedback_consistent": consistent,
    })

print(json.dumps(feedback_integrity, indent=2))
```

## 36. Closed-Loop Fault Scenario

Introduce an authorization failure to verify that an unauthorized action does not change the represented state.


```python
# Phase 5 — authorization failure / safe response
fault_state = copy.deepcopy(closed_loop_state)
fault_decision = make_decision(fault_state, environment)

fault_authorization = {
    "decision": fault_decision["decision"],
    "authorized": False,
    "authorization_mode": "simulated_rejection",
}

fault_actuation = apply_simulated_actuation(
    fault_state,
    fault_authorization["decision"],
    fault_authorization["authorized"]
)

fault_safe_response = {
    "authorization_rejected": not fault_authorization["authorized"],
    "actuation_executed": fault_actuation["actuated"],
    "state_preserved": fault_actuation["state"] == fault_state,
}

print(json.dumps(fault_safe_response, indent=2))
```

## 37. Degraded / Recovery Scenario

Test a simple degraded condition in which the sensing path cannot provide a required value.

The controlled response is to prevent a normal decision cycle from silently proceeding.


```python
# Phase 5 — degraded sensing scenario
degraded_state = copy.deepcopy(current_state)
degraded_state.pop("soil_moisture_pct", None)

required_feedback = [
    "soil_moisture_pct",
    "water_available_l",
    "crop_stress_index",
]

missing_feedback = [
    key for key in required_feedback
    if key not in degraded_state
]

degraded_response = {
    "scenario": "missing_feedback_value",
    "detected": len(missing_feedback) > 0,
    "missing_values": missing_feedback,
    "normal_decision_allowed": len(missing_feedback) == 0,
    "recommended_mode": "SAFE_DEGRADED_MODE"
    if missing_feedback else "NORMAL"
}

print(json.dumps(degraded_response, indent=2))
```

## 38. Closed-Loop Stability / Resource Observation

Observe whether repeated cycles move the represented state in a controlled direction and whether resources remain within configured bounds.

This is an experimental observation, not a claim of production control stability.


```python
# Phase 5 — simple stability/resource observations
stability_metrics = {
    "cycles_completed": len(cycle_records),
    "final_soil_moisture_pct": current_state["soil_moisture_pct"],
    "initial_soil_moisture_pct": closed_loop_state["soil_moisture_pct"],
    "final_water_available_l": current_state["water_available_l"],
    "initial_water_available_l": closed_loop_state["water_available_l"],
    "final_crop_stress_index": current_state["crop_stress_index"],
    "initial_crop_stress_index": closed_loop_state["crop_stress_index"],
    "water_nonnegative": current_state["water_available_l"] >= 0,
    "soil_moisture_bounded": 0 <= current_state["soil_moisture_pct"] <= 100,
    "stress_bounded": 0 <= current_state["crop_stress_index"] <= 1,
}

print(json.dumps(stability_metrics, indent=2))
```

## 39. Closed-Loop KPI Measurement

## 41. Phase 5 → Phase 6 Readiness Gate

The Phase 5 gate checks whether the controlled closed-loop cycle has been demonstrated sufficiently to proceed to the next phase.

~~~text
Observe
  ↓
Decide
  ↓
Authorize
  ↓
Actuate
  ↓
Changed State
  ↓
Feedback
  ↓
Repeated Cycle
  ↓
Fault / Degraded Response
  ↓
Measurement
  ↓
Evidence
  ↓
Formal Review
~~~


```python
# Phase 5 — high-level readiness gate
phase5_gate = {
    "initial_state_defined": True,
    "observation_executed": len(cycle_records) > 0,
    "decision_executed": all(
        bool(r["decision"]["decision"]) for r in cycle_records
    ),
    "authorization_defined": all(
        "authorized" in r["authorization"] for r in cycle_records
    ),
    "simulated_actuation_executed": all(
        "actuated" in r["actuation"] for r in cycle_records
    ),
    "changed_state_recorded": all(
        "changed_state" in r for r in cycle_records
    ),
    "feedback_executed": all(
        "feedback" in r for r in cycle_records
    ),
    "feedback_integrity": (
        all(item["feedback_consistent"] for item in feedback_integrity)
        if feedback_integrity else True
    ),
    "fault_response_safe": fault_safe_response["state_preserved"],
    "degraded_condition_detected": degraded_response["detected"],
    "kpis_measured": bool(phase5_kpis),
    "evidence_recorded": bool(phase5_evidence),
    "physical_automated_actuation": False,
}

phase5_ready = all(phase5_gate.values())

print("Phase 5 Gate")
for key, value in phase5_gate.items():
    print(f"  {key}: {value}")

print(
    "\nPhase 5 status:",
    "READY FOR FORMAL REVIEW" if phase5_ready else "REVISE"
)
```
