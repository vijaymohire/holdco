# Phase 4 — Open-Loop Simulation / QAI Evaluation

This section is appended to the Phase 3 notebook to preserve notebook lineage.

**Phase 3 remains the workflow-definition baseline.**

Phase 4 extends that workflow into a controlled open-loop experiment:

**Input / Sensing → Processing → Digital Farm Twin Context → Classical / AI / QAI → Optimisation → Recommendation → KPI / Value Measurement → Evidence**

**Pilot boundary:** no automated physical actuation.


## 12. Phase 4 Experiment Configuration

Define one bounded agriculture optimisation problem and its controlled experiment configuration.

The initial implementation uses synthetic data so that the experiment is deterministic and reproducible. Real or historical data can be introduced later without changing the logical workflow.


```python
# Phase 4 — experiment configuration
from dataclasses import dataclass, asdict
from typing import Dict, Any, List
import json
import math
import statistics

@dataclass
class Phase4Config:
    experiment_id: str = "FAEP-P4-OPENLOOP-001"
    crop: str = "vegetable"
    field_area_ha: float = 10.0
    scenario: str = "cold_weather_water_constraint"
    data_realism: str = "synthetic"
    device_realism: str = "virtual"
    actuation_mode: str = "recommendation_only"
    random_seed: int = 42

config = Phase4Config()

print("Phase 4 configuration")
for k, v in asdict(config).items():
    print(f"  {k}: {v}")
```

## 13. Input State / Scenario

The experiment represents a bounded field condition using synthetic sensing and environmental inputs.

The scenario is deliberately small so that the complete sensing-to-recommendation path can be inspected before adding physical assets.


```python
# Phase 4 — synthetic sensing/environment input
sensor_input = {
    "soil_moisture_pct": 24.0,
    "air_temperature_c": -3.0,
    "relative_humidity_pct": 82.0,
    "water_available_l": 18000.0,
    "crop_stress_index": 0.35,
    "equipment_availability_pct": 90.0,
    "inventory_units": 1200.0,
    "demand_units": 1100.0,
    "transport_capacity_units": 1150.0,
}

environment = {
    "frost_risk": 0.72,
    "precipitation_mm": 2.0,
    "forecast_cold_hours": 8.0,
}

print("Synthetic input state:")
for k, v in sensor_input.items():
    print(f"  {k}: {v}")

print("\nEnvironmental scenario:")
for k, v in environment.items():
    print(f"  {k}: {v}")
```

## 14. Sensing / Input Validation

Validate the experiment inputs before constructing the Digital Farm Twin context.


```python
# Phase 4 — input validation
required_inputs = [
    "soil_moisture_pct",
    "air_temperature_c",
    "relative_humidity_pct",
    "water_available_l",
    "crop_stress_index",
    "equipment_availability_pct",
    "inventory_units",
    "demand_units",
    "transport_capacity_units",
]

missing = [k for k in required_inputs if k not in sensor_input]
range_checks = {
    "soil_moisture_pct": 0 <= sensor_input["soil_moisture_pct"] <= 100,
    "relative_humidity_pct": 0 <= sensor_input["relative_humidity_pct"] <= 100,
    "crop_stress_index": 0 <= sensor_input["crop_stress_index"] <= 1,
    "equipment_availability_pct": 0 <= sensor_input["equipment_availability_pct"] <= 100,
}

input_validation = {
    "missing_inputs": missing,
    "range_checks": range_checks,
    "valid": (len(missing) == 0 and all(range_checks.values()))
}

print("Input validation:", input_validation)
```

## 15. Digital Farm Twin Context

Construct the minimum virtual system state required for the experiment.

This is an experimental Twin context, not a claim of a production Digital Twin.


```python
# Phase 4 — minimum Digital Farm Twin context
twin_context = {
    "field": {
        "id": "FIELD-001",
        "area_ha": config.field_area_ha,
        "crop": config.crop,
    },
    "environment": environment,
    "sensing": sensor_input,
    "operational": {
        "equipment_availability_pct": sensor_input["equipment_availability_pct"],
        "inventory_units": sensor_input["inventory_units"],
        "demand_units": sensor_input["demand_units"],
        "transport_capacity_units": sensor_input["transport_capacity_units"],
    },
    "data_realism": config.data_realism,
    "device_realism": config.device_realism,
}

print("Digital Farm Twin context constructed.")
print(json.dumps(twin_context, indent=2))
```

## 16. Classical Baseline

Establish a transparent Classical baseline before evaluating an alternative computational path.

For this first open-loop slice, the baseline selects a simple rule-based recommendation from the observed state. It is intentionally interpretable rather than presented as an optimal production algorithm.


```python
# Phase 4 — transparent Classical baseline
def classical_baseline(state: Dict[str, Any]) -> Dict[str, Any]:
    actions = []
    if state["soil_moisture_pct"] < 30 and state["water_available_l"] > 5000:
        actions.append("prioritise_water_allocation")
    if state["frost_risk"] >= 0.60:
        actions.append("activate_frost_protection_plan")
    if state["inventory_units"] < state["demand_units"]:
        actions.append("prioritise_supply_recovery")

    return {
        "method": "classical_rule_baseline",
        "actions": actions,
        "feasible": True,
    }

baseline_result = classical_baseline(sensor_input)
print(json.dumps(baseline_result, indent=2))
```

## 17. AI / QAI-Compatible Alternative

The Phase 4 architecture permits AI/QAI alternatives, but this initial notebook uses a small deterministic scoring model as a **QAI-compatible experimental placeholder** rather than claiming quantum execution or quantum advantage.

The important Phase 4 objective is to establish the common comparison interface.


```python
# Phase 4 — deterministic alternative scoring model
# This is NOT a quantum-advantage claim and does not require a QPU.

candidate_actions = [
    {"name": "maintain_current_plan", "water_use_l": 0, "risk_reduction": 0.05, "cost": 0},
    {"name": "prioritise_water_allocation", "water_use_l": 3000, "risk_reduction": 0.35, "cost": 120},
    {"name": "activate_frost_protection_plan", "water_use_l": 1000, "risk_reduction": 0.45, "cost": 180},
    {"name": "combined_water_and_frost_plan", "water_use_l": 4000, "risk_reduction": 0.65, "cost": 260},
]

def score_candidate(candidate, state, env):
    risk_weight = 1000.0
    resource_weight = 0.01
    return (
        risk_weight * candidate["risk_reduction"] * (0.5 + env["frost_risk"])
        - resource_weight * candidate["water_use_l"]
        - candidate["cost"]
        - 200.0 * state["crop_stress_index"]
    )

scored_candidates = [
    {**c, "score": score_candidate(c, sensor_input, environment)}
    for c in candidate_actions
]

alternative_result = max(scored_candidates, key=lambda x: x["score"])

print("Alternative computational path:")
print(json.dumps(alternative_result, indent=2))
```

## 18. Common Evaluation

Classical and alternative paths are evaluated using a common interface.

This is the point at which later AI, hybrid QAI, or optional remote quantum implementations can replace the placeholder without changing the experiment contract.


```python
# Phase 4 — common evaluation
classical_actions = set(baseline_result["actions"])

def evaluate_result(result_name, recommendation, estimated_cost):
    return {
        "method": result_name,
        "recommendation": recommendation,
        "estimated_cost": estimated_cost,
        "execution_status": "completed",
    }

classical_eval = evaluate_result(
    "Classical baseline",
    baseline_result["actions"],
    120 if baseline_result["actions"] else 0
)

alternative_eval = evaluate_result(
    "Alternative scoring path",
    [alternative_result["name"]],
    alternative_result["cost"]
)

comparison = {
    "classical": classical_eval,
    "alternative": alternative_eval,
}

print(json.dumps(comparison, indent=2))
```

## 19. Recommendation

The open-loop experiment produces a recommendation only.

No physical actuator is called and no field operation is automatically executed.


```python
# Phase 4 — recommendation boundary
recommendation = {
    "experiment_id": config.experiment_id,
    "recommended_action": alternative_result["name"],
    "reason": "highest experimental score under the configured scenario",
    "estimated_cost": alternative_result["cost"],
    "estimated_water_use_l": alternative_result["water_use_l"],
    "risk_reduction_score": alternative_result["risk_reduction"],
    "control_mode": "human_review_required",
    "physical_actuation": False,
}

print("OPEN-LOOP RECOMMENDATION")
print(json.dumps(recommendation, indent=2))
```

## 20. KPI / Value Measurement

Measure technical and value-oriented outputs from the experiment.

These are **simulation/experimental outputs**, not measured farm savings or production outcomes.


```python
# Phase 4 — KPI and value measurements
baseline_cost = classical_eval["estimated_cost"]
alternative_cost = alternative_eval["estimated_cost"]

kpis = {
    "workflow_completed": True,
    "input_validation_passed": input_validation["valid"],
    "baseline_cost": baseline_cost,
    "alternative_cost": alternative_cost,
    "cost_difference": baseline_cost - alternative_cost,
    "recommended_water_use_l": alternative_result["water_use_l"],
    "risk_reduction_score": alternative_result["risk_reduction"],
    "physical_actuation": False,
    "data_realism": config.data_realism,
    "device_realism": config.device_realism,
}

print("Phase 4 KPIs")
for k, v in kpis.items():
    print(f"  {k}: {v}")
```

## 21. Fault / Edge Scenario

Introduce a controlled missing-input case to confirm that the open-loop pipeline does not silently proceed with invalid state.


```python
# Phase 4 — controlled fault case
faulty_input = dict(sensor_input)
faulty_input.pop("soil_moisture_pct")

fault_missing = [k for k in required_inputs if k not in faulty_input]
fault_result = {
    "scenario": "missing_soil_moisture",
    "detected": len(fault_missing) > 0,
    "missing_inputs": fault_missing,
    "execution_allowed": len(fault_missing) == 0,
}

print("Fault scenario result:")
print(json.dumps(fault_result, indent=2))
```

## 22. Reproducibility / Evidence Record

Record the configuration, input realism, computational paths, recommendation, KPIs and limitations as a single experiment object.


```python
# Phase 4 — experiment evidence record
evidence_record = {
    "experiment_id": config.experiment_id,
    "phase": "Phase 4",
    "purpose": "Open-loop sensing-to-recommendation validation",
    "configuration": asdict(config),
    "input_state": sensor_input,
    "environment": environment,
    "twin_context": twin_context,
    "baseline": baseline_result,
    "alternative_path": alternative_result,
    "recommendation": recommendation,
    "kpis": kpis,
    "fault_test": fault_result,
    "claims": {
        "quantum_advantage": "not demonstrated",
        "production_performance": "not demonstrated",
        "physical_field_savings": "not measured",
        "physical_actuation": "not performed",
    },
}

print("Evidence record created.")
print("Experiment ID:", evidence_record["experiment_id"])
print("Evidence fields:", len(evidence_record))
```
